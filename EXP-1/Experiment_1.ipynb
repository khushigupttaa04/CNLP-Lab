Experiment-1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/drive/MyDrive/CNLP-Lab/EXP -1/employee_information_100.csv")
print(df.head())
```

```python
avg_salary = df.groupby("Department")["Salary"].mean()

print("\nAverage Salary of Each Department")
print(avg_salary)

plt.figure(figsize=(8,5))
avg_salary.plot(kind="bar", color="skyblue")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)

plt.show()
```
